```python
from google.colab import drive
import shutil
import os

# Conectar Google Drive
drive.mount('/content/drive')

# Copiar el dataset al almacenamiento de Colab
origen = "/content/drive/MyDrive/VC/P4/datos"
destino = "/content/datos"

shutil.copytree(origen, destino, dirs_exist_ok=True)

# Comprobar que existe el archivo de configuración
DATASET = "/content/datos/dataset_limpio/data.yaml"

print("¿Existe data.yaml?", os.path.exists(DATASET))

# Mostrar su contenido
with open(DATASET, "r") as archivo:
    print(archivo.read())
```

```python
from ultralytics import YOLO

# Cargar modelo preentrenado
model = YOLO("yolo11n.pt")

# Entrenar
model.train(
    data=DATASET,
    epochs=60,
    imgsz=640,
    batch=4,
    device=0,
    project="/content/drive/MyDrive/VC/P4/salida",
    name="yolo_matriculas"
)
```
