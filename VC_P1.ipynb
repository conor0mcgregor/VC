*Resolución manual (forma estudiante):*

```python
# Creación de tablero de ajedrez (forma manual/estudiante)
import numpy as np
import matplotlib.pyplot as plt

tam = 800
tablero = np.zeros((tam, tam, 1), dtype=np.uint8)

# Cada casilla es de 100x100, con 8 filas y 8 columnas
for i in range(8):
    for j in range(8):
        # Si la suma de fila y columna es par, blanco, si no, negro
        if (i + j) % 2 == 0:
            color = 255
        else:
            color = 0
            
        # Rellenar la casilla iterando pixel a pixel para que sea 'manual'
        for y in range(i*100, (i+1)*100):
            for x in range(j*100, (j+1)*100):
                tablero[y, x] = color

plt.imshow(tablero, cmap='gray')
plt.title('Ajedrez Manual')
plt.axis('off')
plt.show()
```

*Resolución con IA (código más optimizado):*

```python
# Tablero de ajedrez (versión IA más directa y pro)
import numpy as np
import matplotlib.pyplot as plt

# Creamos el patrón básico de 2x2 casillas
patron = np.array([[255, 0], [0, 255]], dtype=np.uint8)

# Usamos np.kron (producto de Kronecker) para escalar cada pixel del patrón a un bloque de 100x100
bloque = np.ones((100, 100), dtype=np.uint8)
tablero_ia = np.kron(np.tile(patron, (4, 4)), bloque)

plt.imshow(tablero_ia, cmap='gray')
plt.title('Ajedrez IA')
plt.axis('off')
plt.show()
```

**Comparativa**: El código manual usa múltiples bucles anidados que en Python son lentos y hace que el código sea más largo, aunque muy fácil de entender paso a paso. El código de la IA utiliza funciones avanzadas de la librería numpy (`np.tile`, `np.kron`) que están altamente optimizadas internamente, logrando el mismo resultado en apenas dos líneas de código.

**TAREA**: Crear una imagen estilo Mondrian (un ejemplo https://www3.gobiernodecanarias.org/medusa/ecoescuela/sa/2017/04/17/descubriendo-a-mondrian/) con las funciones de dibujo de OpenCV.
No hagas uso de herramientas de IA, parte del ejemplo anterior.

```python
# Mondrian de forma manual
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Empezamos con un lienzo blanco
lienzo = np.ones((600, 600, 3), dtype=np.uint8) * 255

# Definimos colores (ojo, OpenCV usa BGR)
rojo = (0, 0, 255)
azul = (255, 0, 0)
amarillo = (0, 255, 255)
negro = (0, 0, 0)

# Dibujamos los bloques de color con rectángulos rellenos (-1)
cv2.rectangle(lienzo, (0, 0), (200, 300), rojo, -1)
cv2.rectangle(lienzo, (200, 0), (600, 400), amarillo, -1)
cv2.rectangle(lienzo, (0, 300), (200, 600), azul, -1)
cv2.rectangle(lienzo, (200, 400), (400, 600), negro, -1)
# (El bloque inferior derecho ya es blanco, no hace falta pintarlo)

# Trazamos las líneas separadoras gruesas típicas de Mondrian
grosor = 10
cv2.line(lienzo, (200, 0), (200, 600), negro, grosor)
cv2.line(lienzo, (0, 300), (200, 300), negro, grosor)
cv2.line(lienzo, (200, 400), (600, 400), negro, grosor)
cv2.line(lienzo, (400, 400), (400, 600), negro, grosor)

# Y el borde exterior
cv2.rectangle(lienzo, (0, 0), (600, 600), negro, grosor)

# Convertimos de BGR a RGB para que matplotlib lo muestre bien
lienzo_rgb = cv2.cvtColor(lienzo, cv2.COLOR_BGR2RGB)
plt.imshow(lienzo_rgb)
plt.axis('off')
plt.show()
```

Pintar círculos en las posiciones del píxel más claro y oscuro de cada fotograma captado por la cámara. ¿Funciona de forma fluida o a saltos? En el segundo caso, ¿podrías acelerarlo?

```python
# Píxel más claro y oscuro (Webcam)
import cv2

vid = cv2.VideoCapture(0)

while True:
    ret, frame = vid.read()
    if not ret:
        break
        
    # Convertimos a grises para facilitar encontrar el min y el max
    gris = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    
    # En vez de usar bucles for que irían a saltos (muy lentos en python),
    # usamos la función minMaxLoc de OpenCV que está escrita en C y es muy rápida.
    # Esto permite que el vídeo vaya fluido y sin lag.
    min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(gris)
    
    # Marcamos el píxel más oscuro con un círculo azul y texto
    cv2.circle(frame, min_loc, 20, (255, 0, 0), 2)
    cv2.putText(frame, 'Oscuro', (min_loc[0]-30, min_loc[1]-30), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 0, 0), 2)
    
    # Marcamos el píxel más claro con un círculo amarillo y texto
    cv2.circle(frame, max_loc, 20, (0, 255, 255), 2)
    cv2.putText(frame, 'Claro', (max_loc[0]-25, max_loc[1]-30), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 255), 2)
    
    cv2.imshow('Buscando min/max', frame)
    
    if cv2.waitKey(20) == 27: # ESC para salir
        break

vid.release()
cv2.destroyAllWindows()
```

TAREA: Llevar a cabo una propuesta propia de pop art.

```python
# Propuesta Pop Art propia: Filtro cuádruple estilo cómic
# Inspirado en las variaciones de color típicas del Pop Art.
import cv2
import numpy as np

vid = cv2.VideoCapture(0)

# Reducimos resolución para que entre bien en la pantalla
w = int(vid.get(cv2.CAP_PROP_FRAME_WIDTH) / 2)
h = int(vid.get(cv2.CAP_PROP_FRAME_HEIGHT) / 2)

while True:
    ret, frame = vid.read()
    if not ret:
        break
        
    frame = cv2.resize(frame, (w, h))
    
    # Separamos canales de la cámara (B, G, R)
    b, g, r = cv2.split(frame)
    
    # Cuadro 1 (Arriba Izq): Tonos azulados y alto contraste
    c1 = cv2.merge([cv2.add(b, 50), g, r])
    
    # Cuadro 2 (Arriba Der): Colores invertidos (estilo negativo fotográfico)
    c2 = cv2.bitwise_not(frame)
    
    # Cuadro 3 (Abajo Izq): Intercambiamos verde por rojo (colores surrealistas)
    c3 = cv2.merge([b, r, g])
    
    # Cuadro 4 (Abajo Der): Efecto silueta bicolor (amarillo y morado)
    gris = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    _, mascara = cv2.threshold(gris, 100, 255, cv2.THRESH_BINARY)
    c4 = cv2.cvtColor(mascara, cv2.COLOR_GRAY2BGR)
    # Pintamos la parte blanca de amarillo
    c4[mascara == 255] = [0, 255, 255]
    # Y la negra de morado (BGR)
    c4[mascara == 0] = [128, 0, 128]
    
    # Montamos el collage final apilando horizontal y luego verticalmente
    fila1 = np.hstack((c1, c2))
    fila2 = np.hstack((c3, c4))
    pop_art = np.vstack((fila1, fila2))
    
    cv2.imshow('Mi Pop Art', pop_art)
    
    if cv2.waitKey(20) == 27: # ESC
        break
        
vid.release()
cv2.destroyAllWindows()
```
